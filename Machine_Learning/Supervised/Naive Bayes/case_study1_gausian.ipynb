```python
import pandas as pd
import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
```

```python
data = {
    "Temperature": [98.6, 99.1, 100.2, 101.5, 97.9, 102.1, 98.4, 100.8],
    "BP":          [120, 130, 140, 150, 118, 155, 122, 145],
    "Glucose":     [85, 90, 110, 130, 80, 135, 88, 125],
    "Disease":     [0, 0, 1, 1, 0, 1, 0, 1]
}

df = pd.DataFrame(data)
df
```

```python
X = df[["Temperature", "BP", "Glucose"]]
y = df["Disease"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

model = GaussianNB()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:",accuracy_score(y_test, y_pred))
print("Classifivation_report",classification_report(y_test, y_pred))
```
