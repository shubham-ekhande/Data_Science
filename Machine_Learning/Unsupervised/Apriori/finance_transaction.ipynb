```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_csv("finance_transactions.csv")
df.head()
```

```python
#converting into list
t = df['Items'].apply(lambda x: x.split(', ')).tolist()
```

```python
#one hot encoding
te = TransactionEncoder()
te_arr = te.fit(t).transform(t)
df_encoded = pd.DataFrame(te_arr, columns = te.columns_)
```

```python
#apply apriori
freq = apriori(df_encoded, min_support=0.2, use_colnames=True)
```

```python
freq
```

```python
#Gnerate Rules
rules = association_rules(freq, metric='confidence', min_threshold=0.4)
```
