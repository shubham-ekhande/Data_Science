```python
import pandas as pd
```

```python
df = pd.read_csv("amazon.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
"""## Some rows cotain symbols or commas converting them into int or float
df['actual_price'] = (
    df['actual_price']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)

df['discounted_price'] = (
    df['discounted_price']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)"""
```

```python
#Converting str to numeric
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')
df['actual_price'] = pd.to_numeric(df['actual_price'], errors='coerce')
df['discounted_price'] = pd.to_numeric(df['discounted_price'], errors='coerce')
```

```python
# checking null values
df[['rating', 'rating_count', 'actual_price', 'discounted_price']].isna().sum()
```

```python
# filling null values
df['rating'] = df['rating'].fillna(df['rating'].median())
df['rating_count'] = df['rating_count'].fillna(df['rating_count'].median())
df['actual_price'] = df['actual_price'].fillna(df['actual_price'].median())
df['discounted_price'] = df['discounted_price'].fillna(df['discounted_price'].median())
```

```python
# creating purchased column
df['purchased'] = ((df['rating'] >= 4) & (df['rating_count'] >= 500)).astype(int)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
X = df[['rating', 'rating_count', 'actual_price', 'discounted_price']]
y = df['purchased']
```

```python
#scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X, y, test_size=0.3)
```

```python
model = KNeighborsClassifier(n_neighbors = 5, metric = 'euclidean')
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
## Accuracy score 
print(f"Accuracy score: {accuracy_score(y_test,y_pred)}")
```

```python
## Example
test =pd.DataFrame({
        'rating' : [4.3],
        'rating_count' : [2000],
        'actual_price' : [650],
        'discounted_price' : [400]
})

p = model.predict(test)[0]
if p == 1:
    print("Customer is Likely to Purchase the product")

else:
    print("Customer is not Likely to purchase the product")
```
