```python
import pandas as pd
import umap.umap_ as umap
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv("text_embeddings_50d.csv")
X = df.drop("topic", axis=1)
y = df["topic"]
```

```python
# Apply UMAP
reducer = umap.UMAP(n_neighbors=15, min_dist=0.1, random_state=42)
X_2d = reducer.fit_transform(X)

# Plot
plt.figure()
for topic in y.unique():
    idx = y == topic
    plt.scatter(X_2d[idx, 0], X_2d[idx, 1], label=topic)

plt.legend()
plt.title("UMAP Visualization of Text Embeddings")
plt.show()
```
