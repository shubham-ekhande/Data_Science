```python
import numpy as np
import pandas as pd
from sklearn.neighbors import NearestNeighbors
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("shop_locations_dbscan.csv")
```

```python
df.head()
```

```python
X = df.values
```

```python
plt.scatter(X[:,0], X[:,1])
plt.title("Shop Locations (Raw GPS Data)")
plt.xlabel("Latitude")
plt.ylabel("Longitude")
plt.show()
```

```python
#finding epsilon
n = NearestNeighbors(n_neighbors=5)
n_a = n.fit(X)
d,_ = n_a.kneighbors(X)
```

```python
dis = np.sort(d[:,4])
```

```python
plt.plot(dis)
plt.title("k-Distance Graph (Shop Dataset)")
plt.xlabel("Sorted Shops")
plt.ylabel("5th Nearest Neighbor Distance")
plt.show()
```

```python
db = DBSCAN(eps=0.003, min_samples=5)
labels = db.fit_predict(X)
```

```python
plt.scatter(X[:,0], X[:,1], c=labels)
plt.title("DBSCAN Result: Market Areas & Isolated Shops")
plt.xlabel("Latitude")
plt.ylabel("Longitude")
plt.show()
```
